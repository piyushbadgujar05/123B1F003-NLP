```python
import re
import numpy as np
from collections import defaultdict
```

```python
text = """machine learning is a field of artificial intelligence that focuses on teaching computers to learn from data machine learning algorithms build models based on sample data deep learning is a subset of machine learning natural language processing helps computers understand human language artificial intelligence is transforming the world"""

D = 5
window_size = 2
epochs = 100
learning_rate = 0.01
```

```python
# Lowercase
text = text.lower()

# Remove punctuation
text = re.sub(r'[^a-zA-Z\s]', '', text)

# Tokenize
words = text.split()

print("Words:", words[:20])
```

```python
vocab = list(set(words))
V = len(vocab)

word_to_index = {word: i for i, word in enumerate(vocab)}
index_to_word = {i: word for word, i in word_to_index.items()}

print("Vocabulary Size:", V)
```

```python
training_data = []

for i, word in enumerate(words):
    target_idx = word_to_index[word]

    for j in range(-window_size, window_size + 1):
        if j != 0 and 0 <= i + j < len(words):
            context_idx = word_to_index[words[i + j]]
            training_data.append((target_idx, context_idx))

print("Sample Training Pairs:", training_data[:10])
print("Total Pairs:", len(training_data))
```

```python
np.random.seed(42)

W1 = np.random.randn(V, D) * 0.01
W2 = np.random.randn(D, V) * 0.01
```

```python
def softmax(x):
    exp_x = np.exp(x - np.max(x))
    return exp_x / np.sum(exp_x)

def one_hot(index, size):
    vec = np.zeros(size)
    vec[index] = 1
    return vec
```

```python
for epoch in range(epochs):
    total_loss = 0

    for target_idx, context_idx in training_data:


        x = one_hot(target_idx, V)


        h = np.dot(W1.T, x)
        u = np.dot(W2.T, h)
        y_pred = softmax(u)


        loss = -np.log(y_pred[context_idx] + 1e-9)
        total_loss += loss


        e = y_pred.copy()
        e[context_idx] -= 1

        dW2 = np.outer(h, e)
        dW1 = np.outer(x, np.dot(W2, e))


        W1 -= learning_rate * dW1
        W2 -= learning_rate * dW2


    total_loss /= len(training_data)

    if (epoch + 1) % 10 == 0:
        print(f"Epoch {epoch+1}, Loss: {total_loss:.4f}")
```
