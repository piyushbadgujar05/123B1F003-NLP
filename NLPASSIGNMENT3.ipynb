```python
import re
from collections import defaultdict, Counter
```

```python
def tokenize(text):
    text = text.lower()
    text = re.sub(r"[^a-z\s]", "", text)
    return text.split()
```

```python
class NGramLanguageModel:
    def __init__(self, n):
        self.n = n
        self.ngrams = defaultdict(Counter)

    def train(self, text):
        tokens = tokenize(text)
        for i in range(len(tokens) - self.n + 1):
            context = tuple(tokens[i:i + self.n - 1])
            next_word = tokens[i + self.n - 1]
            self.ngrams[context][next_word] += 1

    def predict_next(self, context):
        context = tuple(context.lower().split())
        if context not in self.ngrams:
            return "No prediction available"
        return self.ngrams[context].most_common(1)[0][0]

    def next_word_probabilities(self, context):
        context = tuple(context.lower().split())
        if context not in self.ngrams:
            return {}
        total = sum(self.ngrams[context].values())
        return {word: count / total for word, count in self.ngrams[context].items()}
```

```python
corpus = """
Natural language processing is a field of artificial intelligence.
Natural language processing allows machines to understand human language.
Machine learning is a part of artificial intelligence.
Deep learning is a subset of machine learning.
Language models are used in natural language processing.
"""
```

```python
model = NGramLanguageModel(n=2)
model.train(corpus)

print("Next word prediction:", model.predict_next("natural"))
print("Probabilities:", model.next_word_probabilities("natural"))
```

```python
import nltk
from nltk.lm import MLE
from nltk.lm.preprocessing import padded_everygram_pipeline
from nltk.tokenize import word_tokenize
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
corpus_text= [
    "Natural language processing is a field of artificial intelligence",
    "Natural language processing allows machines to understand human language",
    "Machine learning is a part of artificial intelligence",
    "Deep learning is a subset of machine learning",
    "Language models are used in natural language processing"
]
```

```python
tokenized_text = [word_tokenize(sentence.lower()) for sentence in corpus_text]
```

```python
n = 2
train_data, padded_vocab = padded_everygram_pipeline(n, tokenized_text)
```

```python
model = MLE(n)
model.fit(train_data, padded_vocab)
```

```python
def predict_next_word(model, context):
    """
    Predicts the most probable next word
    while ignoring </s> token
    """
    context = tuple(context)

    counts = model.context_counts(context)

    if not counts:
        return "No prediction"

    # Remove end-of-sentence token
    candidates = {
        word: count
        for word, count in counts.items()
        if word != '</s>'
    }

    if not candidates:
        return "No prediction"

    return max(candidates, key=candidates.get)
```

```python
context = ["natural"]
prediction = predict_next_word(model, context)
```

```python
print("Context:", " ".join(context))
print("Next word prediction:", prediction)
```
